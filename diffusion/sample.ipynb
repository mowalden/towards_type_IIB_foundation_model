```python
from pathlib import Path
import sys

REPO_ROOT = Path("/content/towards_type_IIB_foundation_model")
sys.path.insert(0, str(REPO_ROOT))
```

```python
import torch
from diffusion.src.pipeline import MODES, sample_and_save

NUM_SAMPLES = 1000  # Per geometry; set to 2 for a quick smoke test.
BATCH_SIZE = 64     # Lower if GPU memory is limited.
SEED = 90261001
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
DISCR = 10.0       # Must match --discr when evaluating.
OUTPUT_DIR = REPO_ROOT / "diffusion/samples"

# None uses the checkpoint's evaluation defaults: confidence order,
# temperature 0.4; guidance 2.0 for conditional models, 1.0 unconditional.
SAMPLER = None     # Optional: "random", "confidence", or "maskgit".
TEMPERATURE = None
GUIDANCE_SCALE = None

for filename, _, _ in MODES.values():
    checkpoint = REPO_ROOT / "diffusion/checkpoints" / filename
    if not checkpoint.is_file():
        raise FileNotFoundError(checkpoint)
print(f"Device: {DEVICE}; {NUM_SAMPLES} samples for each of 20 geometries")


def generate(mode, num_samples=NUM_SAMPLES):
    path = sample_and_save(
        mode,
        output=OUTPUT_DIR / f"diffusion_{mode}_samples.p",
        num_samples=num_samples,
        batch_size=BATCH_SIZE,
        seed=SEED,
        device=DEVICE,
        sampler=SAMPLER,
        temperature=TEMPERATURE,
        guidance_scale=GUIDANCE_SCALE,
        discr=DISCR,
    )
    print(f"Saved {path} and {path.with_suffix('.json')}")
    return path
```

```python
nflux_file = generate(mode="nflux", num_samples=1000)
```

```python
w0_file = generate("w0")
```

```python
gs_inv_file = generate("gs_inv")
```

```python
unconditional_file = generate("unconditional")
```
